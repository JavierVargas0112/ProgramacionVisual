```python
import numpy as np
import matplotlib.pyplot as plt
import imageio
import os

# Crear figura 2D: triángulo
triangle = np.array([
    [0, 1, 0.5, 0],  # x
    [0, 0, 1, 0],    # y
    [1, 1, 1, 1]     # homogénea
])


num_frames = 60
gif_frames = []
output_folder = "frames"
os.makedirs(output_folder, exist_ok=True)


for t in range(num_frames):
    theta = 2 * np.pi * t / num_frames  # rotación
    scale = 1 + 0.5 * np.sin(2 * np.pi * t / num_frames)  # escala oscilante
    tx = 2 * np.cos(2 * np.pi * t / num_frames)  # traslación en x
    ty = 2 * np.sin(2 * np.pi * t / num_frames)  # traslación en y

    # Matrices de transformación
    S = np.array([
        [scale, 0, 0],
        [0, scale, 0],
        [0, 0, 1]
    ])
    R = np.array([
        [np.cos(theta), -np.sin(theta), 0],
        [np.sin(theta),  np.cos(theta), 0],
        [0, 0, 1]
    ])
    T = np.array([
        [1, 0, tx],
        [0, 1, ty],
        [0, 0, 1]
    ])

    # Transformación total: primero escala, luego rota, luego traslada
    transform = T @ R @ S
    transformed_triangle = transform @ triangle

    # Dibujar
    fig, ax = plt.subplots()
    ax.plot(transformed_triangle[0], transformed_triangle[1], 'bo-')
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.axis('off')

    # Guardar frame
    filename = f"{output_folder}/frame_{t:03d}.png"
    plt.savefig(filename, bbox_inches='tight', pad_inches=0)
    plt.close()
    gif_frames.append(imageio.v2.imread(filename))

# Exportar a GIF
imageio.mimsave('animacion.gif', gif_frames, duration=0.05)

```
